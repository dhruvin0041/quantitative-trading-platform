## 2. Dynamic Triple Barrier Labeling
Instead of predicting simple returns, we use Triple Barrier Labeling introduced by Marcos López de Prado. The barriers adapt to current volatility (ATR).

## 3. Market Regime Detection
Models behave differently in Bull vs. Bear markets. We classify regimes using the SPY 50-day and 200-day moving averages and adjust our signal conviction thresholds accordingly.

## 4. Model Architectures & Meta-Ensemble
Our consensus model evaluates signals from:
1. **Deep Learning Fusion Network** (CNN + LSTM + Transformer + FinBERT)
2. **Gradient Boosting Agents** (XGBoost, LightGBM)
3. **Reinforcement Learning** (DQN)
4. **State-of-the-Art Transformers** (TFT, PatchTST, Informer)

The Meta-Ensemble dynamically weights these models based on recent 30-day out-of-sample accuracy.

## 5. Walk-Forward Optimization (WFO)
We prevent overfitting by continuously rolling our train/test windows. The backtester strictly trades on the out-of-sample forward window.
